```python
"""
Script SIMPLIFICADO para limpieza de datos de texto
No requiere descargas externas, usa solo expresiones regulares
"""

import re
import unicodedata


class LimpiadorTextoSimple:
    """Clase para limpiar y procesar texto sin dependencias externas"""

    def __init__(self):
        # Stopwords comunes en español (lista básica)
        self.stopwords_es = {
            "el",
            "la",
            "de",
            "que",
            "y",
            "a",
            "en",
            "un",
            "ser",
            "se",
            "no",
            "haber",
            "por",
            "con",
            "su",
            "para",
            "como",
            "estar",
            "tener",
            "le",
            "lo",
            "todo",
            "pero",
            "más",
            "hacer",
            "o",
            "poder",
            "decir",
            "este",
            "ir",
            "otro",
            "ese",
            "la",
            "si",
            "me",
            "ya",
            "ver",
            "porque",
            "dar",
            "cuando",
            "él",
            "muy",
            "sin",
            "vez",
            "mucho",
            "saber",
            "qué",
            "sobre",
            "mi",
            "alguno",
            "mismo",
            "yo",
            "también",
            "hasta",
            "año",
            "dos",
            "querer",
            "entre",
            "así",
            "primero",
            "desde",
            "grande",
            "eso",
            "ni",
            "nos",
            "llegar",
            "pasar",
            "tiempo",
            "ella",
            "sí",
            "día",
            "uno",
            "bien",
            "poco",
            "deber",
            "entonces",
            "poner",
            "cosa",
            "tanto",
            "hombre",
            "parecer",
            "nuestro",
            "tan",
            "donde",
            "ahora",
            "parte",
            "después",
            "vida",
            "quedar",
            "siempre",
            "creer",
            "hablar",
            "llevar",
            "dejar",
            "nada",
            "cada",
            "seguir",
            "menos",
            "nuevo",
            "encontrar",
            "algo",
            "solo",
            "decir",
            "salir",
            "volver",
            "tomar",
            "conocer",
            "vivir",
            "sentir",
            "tratar",
            "mirar",
            "contar",
            "empezar",
            "esperar",
            "buscar",
            "existir",
            "entrar",
            "trabajar",
            "escribir",
            "perder",
            "producir",
            "ocurrir",
            "entender",
            "pedir",
            "recibir",
            "recordar",
            "terminar",
            "permitir",
            "aparecer",
            "conseguir",
            "comenzar",
            "servir",
            "sacar",
            "necesitar",
            "mantener",
            "resultar",
            "leer",
            "caer",
            "cambiar",
            "presentar",
            "crear",
            "abrir",
            "considerar",
            "oír",
            "acabar",
            "mil",
            "contra",
            "cual",
            "durante",
            "ellos",
            "las",
            "los",
            "mis",
            "nosotros",
            "sus",
            "te",
            "ti",
            "tu",
            "tus",
            "vosotros",
            "vuestra",
            "vuestro",
        }

        # Diccionario de correcciones comunes (puedes expandirlo)
        self.correcciones = {
            "hola": "hola",
            "q": "que",
            "xq": "porque",
            "pq": "porque",
            "tmb": "también",
            "tb": "también",
            "x": "por",
            "d": "de",
            "k": "que",
            "bn": "bien",
            "mxo": "mucho",
            "msj": "mensaje",
            "msg": "mensaje",
            "jajaja": "jaja",
            "jajajaja": "jaja",
            "jjjjj": "jaja",
            "jeje": "jaja",
            "jiji": "jaja",
            "ajaj": "jaja",
        }

    def a_minusculas(self, texto):
        """Convierte texto a minúsculas"""
        return texto.lower()

    def eliminar_acentos(self, texto):
        """Elimina acentos del texto (opcional)"""
        return "".join(
            c
            for c in unicodedata.normalize("NFD", texto)
            if unicodedata.category(c) != "Mn"
        )

    def eliminar_urls(self, texto):
        """Elimina URLs del texto"""
        return re.sub(r"http\S+|www\S+|https\S+", "", texto, flags=re.MULTILINE)

    def eliminar_emails(self, texto):
        """Elimina emails del texto"""
        return re.sub(r"\S+@\S+", "", texto)

    def eliminar_menciones_hashtags(self, texto):
        """Elimina menciones (@usuario) y hashtags (#tema)"""
        texto = re.sub(r"@\w+", "", texto)
        texto = re.sub(r"#\w+", "", texto)
        return texto

    def eliminar_numeros(self, texto):
        """Elimina números del texto"""
        return re.sub(r"\d+", "", texto)

    def eliminar_caracteres_especiales(self, texto, mantener_espacios=True):
        """
        Elimina caracteres especiales y puntuación
        Mantiene letras, números (si no se eliminaron antes) y opcionalmente espacios
        """
        if mantener_espacios:
            # Mantener solo letras, números y espacios
            texto = re.sub(r"[^a-záéíóúñüA-ZÁÉÍÓÚÑÜ\s]", "", texto)
        else:
            texto = re.sub(r"[^a-záéíóúñüA-ZÁÉÍÓÚÑÜ]", " ", texto)
        return texto

    def eliminar_espacios_extra(self, texto):
        """Elimina espacios múltiples y limpia el texto"""
        return " ".join(texto.split()).strip()

    def corregir_palabras_comunes(self, texto):
        """Corrige palabras comunes mal escritas"""
        palabras = texto.split()
        palabras_corregidas = [
            self.correcciones.get(palabra, palabra) for palabra in palabras
        ]
        return " ".join(palabras_corregidas)

    def eliminar_palabras_repetidas(self, texto):
        """Elimina palabras que se repiten consecutivamente"""
        palabras = texto.split()
        resultado = []
        anterior = None
        for palabra in palabras:
            if palabra != anterior:
                resultado.append(palabra)
                anterior = palabra
        return " ".join(resultado)

    def eliminar_stopwords(self, texto):
        """Elimina palabras vacías (stopwords)"""
        palabras = texto.split()
        palabras_filtradas = [
            palabra
            for palabra in palabras
            if palabra not in self.stopwords_es and len(palabra) > 2
        ]
        return " ".join(palabras_filtradas)

    def limpiar_texto_completo(
        self,
        texto,
        remover_acentos=False,
        remover_stopwords=False,
        corregir_palabras=True,
    ):
        """
        Aplica todas las transformaciones de limpieza

        Args:
            texto: texto a limpiar
            remover_acentos: si se deben eliminar acentos
            remover_stopwords: si se deben eliminar stopwords
            corregir_palabras: si se deben corregir palabras comunes

        Returns:
            texto limpio
        """
        if not isinstance(texto, str):
            return ""

        # 1. Convertir a minúsculas
        texto = self.a_minusculas(texto)

        # 2. Eliminar URLs
        texto = self.eliminar_urls(texto)

        # 3. Eliminar emails
        texto = self.eliminar_emails(texto)

        # 4. Eliminar menciones y hashtags
        texto = self.eliminar_menciones_hashtags(texto)

        # 5. Eliminar números
        texto = self.eliminar_numeros(texto)

        # 6. Corregir palabras comunes (antes de eliminar caracteres especiales)
        if corregir_palabras:
            texto = self.corregir_palabras_comunes(texto)

        # 7. Eliminar caracteres especiales
        texto = self.eliminar_caracteres_especiales(texto)

        # 8. Eliminar acentos (opcional)
        if remover_acentos:
            texto = self.eliminar_acentos(texto)

        # 9. Eliminar palabras repetidas
        texto = self.eliminar_palabras_repetidas(texto)

        # 10. Eliminar espacios extra
        texto = self.eliminar_espacios_extra(texto)

        # 11. Eliminar stopwords (opcional)
        if remover_stopwords:
            texto = self.eliminar_stopwords(texto)

        return texto


def limpiar_X_train(
    X_train,
    remover_acentos=False,
    remover_stopwords=False,
    corregir_palabras=True,
    verbose=True,
):
    """
    Limpia un conjunto de datos X_train

    Args:
        X_train: lista, array o Series con textos
        remover_acentos: si se deben eliminar acentos
        remover_stopwords: si se deben eliminar stopwords
        corregir_palabras: si se deben corregir palabras comunes
        verbose: mostrar progreso

    Returns:
        lista con textos limpios
    """
    limpiador = LimpiadorTextoSimple()

    X_train_limpio = []
    total = len(X_train)

    for i, texto in enumerate(X_train):
        texto_limpio = limpiador.limpiar_texto_completo(
            texto,
            remover_acentos=remover_acentos,
            remover_stopwords=remover_stopwords,
            corregir_palabras=corregir_palabras,
        )
        X_train_limpio.append(texto_limpio)

        # Mostrar progreso
        if verbose and (i + 1) % 100 == 0:
            print(f"✓ Procesados {i + 1}/{total} textos ({(i + 1) / total * 100:.1f}%)")

    if verbose:
        print(f"✓ Limpieza completada: {total} textos procesados")

    return X_train_limpio


def mostrar_comparacion(original, limpio, n=5):
    """Muestra una comparación entre textos originales y limpios"""
    print("\n" + "=" * 80)
    print("COMPARACIÓN DE TEXTOS")
    print("=" * 80)

    for i in range(min(n, len(original))):
        print(f"\n[{i + 1}] ORIGINAL:")
        print(
            f"    {original[i][:100]}..."
            if len(original[i]) > 100
            else f"    {original[i]}"
        )
        print(f"    LIMPIO:")
        print(
            f"    {limpio[i][:100]}..." if len(limpio[i]) > 100 else f"    {limpio[i]}"
        )


# ============= EJEMPLO DE USO =============

if __name__ == "__main__":
    # Datos de ejemplo - Reemplaza con tus datos reales
    X_train = [
        "Hola!!! Este es un EJEMPLO de texto con URLs https://ejemplo.com y números 123",
        "Texto con @usuario y #hashtags xq son comunes en redes sociales 😊",
        "Email: usuario@ejemplo.com y caracteres especiales $$$ ### !!! ???",
        "   Texto   con    espacios    múltiples   y  palabras  repetidas  repetidas  repetidas",
        "Q bueno k esto funcione bn, tmb con abreviaturas x corregir jajajaja",
        "MAYÚSCULAS mezcladas con minúsculas y Números 456 789",
        "Àcentós y ñ eñes para probar la limpieza completa",
        "www.sitio.com http://otro-sitio.es son URLs q eliminar",
    ]

    print("🔹" * 40)
    print("LIMPIADOR DE TEXTO - PYTHON")
    print("🔹" * 40)

    print(f"\nTotal de textos en X_train: {len(X_train)}")

    # OPCIÓN 1: Limpieza estándar (recomendada)
    print("\n" + "=" * 80)
    print("OPCIÓN 1: LIMPIEZA ESTÁNDAR")
    print("=" * 80)

    X_train_limpio = limpiar_X_train(
        X_train,
        remover_acentos=False,  # Mantener acentos
        remover_stopwords=False,  # Mantener stopwords
        corregir_palabras=True,  # Corregir abreviaturas
        verbose=True,
    )

    mostrar_comparacion(X_train, X_train_limpio, n=5)

    # OPCIÓN 2: Limpieza agresiva (para modelos que no requieren stopwords)
    print("\n\n" + "=" * 80)
    print("OPCIÓN 2: LIMPIEZA AGRESIVA (SIN STOPWORDS)")
    print("=" * 80)

    X_train_agresivo = limpiar_X_train(
        X_train,
        remover_acentos=True,  # Eliminar acentos
        remover_stopwords=True,  # Eliminar stopwords
        corregir_palabras=True,
        verbose=False,
    )

    mostrar_comparacion(X_train, X_train_agresivo, n=3)

    # Estadísticas
    print("\n" + "=" * 80)
    print("ESTADÍSTICAS")
    print("=" * 80)

    palabras_original = sum(len(t.split()) for t in X_train)
    palabras_limpio = sum(len(t.split()) for t in X_train_limpio)
    palabras_agresivo = sum(len(t.split()) for t in X_train_agresivo)

    print(f"Palabras totales (original):  {palabras_original}")
    print(
        f"Palabras totales (limpio):    {palabras_limpio} ({palabras_limpio / palabras_original * 100:.1f}%)"
    )
    print(
        f"Palabras totales (agresivo):  {palabras_agresivo} ({palabras_agresivo / palabras_original * 100:.1f}%)"
    )

    print(f"\nPromedio palabras/texto:")
    print(f"  Original:  {palabras_original / len(X_train):.2f}")
    print(f"  Limpio:    {palabras_limpio / len(X_train_limpio):.2f}")
    print(f"  Agresivo:  {palabras_agresivo / len(X_train_agresivo):.2f}")

    # Guardar resultados (opcional)
    print("\n" + "=" * 80)
    print("PARA GUARDAR LOS RESULTADOS:")
    print("=" * 80)
    print("""
    import pandas as pd
    
    # Crear DataFrame con resultados
    df = pd.DataFrame({
        'texto_original': X_train,
        'texto_limpio': X_train_limpio
    })
    
    # Guardar a CSV
    df.to_csv('datos_limpios.csv', index=False, encoding='utf-8')
    print("✓ Guardado en 'datos_limpios.csv'")
    """)
```
